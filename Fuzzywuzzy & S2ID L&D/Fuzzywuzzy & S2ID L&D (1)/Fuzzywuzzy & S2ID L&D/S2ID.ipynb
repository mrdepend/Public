```python
import pandas as pd
import s2sphere
```

```python
df_shp = pd.read_excel('rawmerchantshp.xlsx', index = False)
df_shp.head()
```

```python
#generating s2id from lat long
#changing df format to list
lat = df_shp.gps_latitude_shp.values.tolist()
lon = df_shp.gps_longitude_shp.values.tolist()

s2id = [] 

cellId1 = []
cellId2 = []
cellId3 = []

for i in range(0,len(lat)):
    p1 = s2sphere.LatLng.from_degrees(lat[i], lon[i])
    cell = s2sphere.CellId.from_lat_lng(p1)
    cell11 = cell.parent(11)
    c11id = str(cell11.id())
    s2id.append(cell11.id())
    #print(cid)
    ##cid is a 19 digit number so python storing it as Object, not number
    ##So I am converting it into 3 numbers
    cellId1.append(int(c11id[:6]))
    cellId2.append(int(c11id[6:12]))
    cellId3.append(int(c11id[12:19]))
    
se = pd.Series(s2id)
#df = se.to_frame()
#df.info()
#df.head()
df_shp['s2id11'] = se.values

se = pd.Series(cellId1)
df_shp['s211id1'] = se.values

se = pd.Series(cellId2)
df_shp['s211id2'] = se.values

se = pd.Series(cellId3)
df_shp['s211id3'] = se.values

df_shp = df_shp.loc[:, ['merchant_id_shp','merchant_name_shp','gps_latitude_shp','gps_longitude_shp','outlet_id_shp','outlet_name_shp','s2id11','s211id1','s211id2','s211id3']]
df_shp.head()
```
