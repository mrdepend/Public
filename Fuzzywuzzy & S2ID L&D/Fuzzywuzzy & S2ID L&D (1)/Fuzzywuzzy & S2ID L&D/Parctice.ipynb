```python
df = pd.read_excel('merchant_checking.xlsx')
df.head(10)
```

```python
df.info()
```

```python
#applying fuzzy wuzzy to dataframe

#changing data frame to list
outlet_name_shp = df.outlet_name_shp.values.tolist()
merchant_name_comp = df.merchant_name_comp.values.tolist()

#using 2 ratio to compare full part name with partial part
ratio = []
partial_ratio = []
token_sort_ratio = []
token_set_ratio = []

for i in range(0,len(outlet_name_shp)):
    r1 = fuzz.ratio(outlet_name_shp[i].lower(), merchant_name_comp[i].lower())
    r2 = fuzz.partial_ratio(outlet_name_shp[i].lower(), merchant_name_comp[i].lower())
    r3 = fuzz.token_sort_ratio(outlet_name_shp[i], merchant_name_comp[i])
    r4 = fuzz.token_set_ratio(outlet_name_shp[i], merchant_name_comp[i])
    ratio.append(r1)
    partial_ratio.append(r2)
    token_sort_ratio.append(r3)
    token_set_ratio.append(r4)
```

```python
df['ratio'] = pd.Series(ratio)
df['partial_ratio'] = pd.Series(partial_ratio)
df['token_sort_ratio'] = pd.Series(token_sort_ratio)
df['token_set_ratio'] = pd.Series(token_set_ratio)
```
